````python
import kagglehub

# --- VARIABLES ---
# 1. The Kaggle dataset handle
dataset_handle = "mustafakeser4/sap-dataset-bigquery-dataset"
dataset_path = "vbak.csv" #we will just grab the sales headers 

# 2. The local folder path where you want the files saved
destination_folder = "../public/datasets/"
# -----------------

# Download the entire dataset to your folder
downloaded_path = kagglehub.dataset_download(
    dataset_handle,
    dataset_path,
    output_dir=destination_folder,
    force_download=True
)

print(f"Dataset successfully downloaded to: {downloaded_path}")
````

````python
import os
import pandas as pd

# --- VARIABLES ---
file_path = "../public/datasets/vbak.csv"
# -----------------

  # Loads the full DataFrame safely for just this one table
df = pd.read_csv(file_path, low_memory=False)
print(f"Successfully loaded ({df.shape[0]:,} rows)")
````
